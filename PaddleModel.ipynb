```python
class PaddleModel(nn.Module):
    def __init__(self,inpF,outp):
        super().__init__()
        self.layers=nn.Sequential(nn.Conv2d(inpF,inpF*2),
                             nn.ReLU(),
                             nn.MaxPool2d(kernel_size=2),
                             nn.Conv2d(inpF*2,inpF*4),
                             nn.ReLU(),
                             nn.AdaptiveAvgPool2d(1),
                             nn.Flatten(),
                             nn.LazyLinear(outp))
        # layers=[]
        # inp=inpF
        
        # for l in range(nConvs):
        #     layers.append(nn.LazyConv2d(inp*2))
        #     layers.append(nn.ReLU())
        #     layers.append(nn.MaxPool2d(kernel_size=2,stride=2))
        # self.layers=nn.Sequential(*layers)
    def forward(self,x):
        return self.layers(x)
```
